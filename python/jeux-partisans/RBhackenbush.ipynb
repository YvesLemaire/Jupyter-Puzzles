## Red Blue Hackenbush

### Règle du jeu

Une position est la donnée   
1. d'un graphe fini $G$ non orienté (les arêtes multiples et les boucles sont autorisées) dont chaque arête est munie d'une couleur bleue ou rouge.
2. et d'un sous-ensemble non vide de l'ensemble des sommets appelé le *sol*. On suppose que chaque sommet est relié par un chemin à au moins un sommet situé sur le sol.

On définit la *charge* d'une arête $e$ de $G$ comme l'ensemble des éléments, sommets ou arêtes, de $G$ qui ne sont plus reliés au sol si on supprime l'arête $e$.

Les joueurs sont appelés *Left* et *Right*.
Chaque joueur, à son tour, choisit une arête, bLeue s'il s'agit de Left, et Rouge pour Right, et la supprime du graphe ; il supprime aussi chaque élément de la charge
de cette arête. Le joueur qui ne peut plus jouer (il n'y a plus d'arête de sa couleur) a perdu.

Dans la figure suivante, le sol est la droite en pointillés.

### L'anneau des nombres dyadiques

Soit $\mathbf D=\{p/2^q\,,\,p\in\mathbf Z, q\in \mathbf N\}$ l'anneau des rationnels *dyadiques*. On note aussi $\overline{\mathbf D}:=\mathbf D\cup\{\emptyset\}$.  
Pour $x,y\in \overline{\mathbf D}$, tel que $x=\emptyset$, ou $y=\emptyset$, ou $x,y \in \mathbf D$ et $x<y$, on note 
$]x,y[_{\mathbf D}:=]x,y[\,\cap\,\mathbf D$ en convenant de remplacer $x$ par $-\infty$ si $x=\emptyset$ et $y$ par $+\infty$ si $y=\emptyset$ et
$$
(x|y):=\begin{cases}
\text{l'entier de $]x,y[_{\mathbf D}$ de valeur absolue minimum}&\text{si } ]x,y[_{\mathbf D}\cap\mathbf Z\neq \emptyset,\\
\text{l'élément de $]x,y[_{\mathbf D}$ de dénominateur minimum}&\text{sinon}
\end{cases}
$$
On utilise $\texttt{sympy}$ pour représenter les fractions de $\mathbf D$ et on représente $\emptyset$ par $\texttt{None}$.


```python
from sympy import *

def V(left, right):

    # left et right sont des nombres dyadiques ou None (pour l'ensemble vide) et left < right
    # renvoie (left | right) en appliquant la définition précédente (appelée principe de simplicité ...)

    auxL = lambda l: floor(l) + 1
    auxR = lambda r: r - 1 if denom(r) == 1 else floor(r)

    if left is None:
        if right is None:
            return S(0)
        else:
            return S(0) if right > 0 else auxR(right)
    else:
        if right is None:
            return S(0) if left < 0 else auxL(left)
        else:
            h = auxL(left)
            k = auxR(right) 
            if k >= h:
                return h if h >= 0 else k if k <= 0 else 0
            else: 
                d, l, r = 1, left, right
                while h != k:
                    d *= 2
                    l *= 2
                    r *= 2
                    h = auxL(l)
                    k = auxR(r)
                return S(h) / d
```

### Valeur d'une position

On définit récursivement la *valeur*  $v=V(G)\in\mathbf D$ d'une position $G$ de RB-Hackenbush de la manière suivante :

A chaque arête  $e$ de $G$ on associe la valeur $v_e$ de la position obtenue en jouant $e$ dans la position $G$. On pose  
$v^L:=\max\{v_e\,,\,e \text{ arête bleue}\}$ ou $\emptyset$ s'il n'y a pas d'arête bleue,   
$v^R:=\min\{v_e\,,\,e \text{ arête rouge}\}$ ou $\emptyset$ s'il n'y a pas d'arête rouge et  
$v:=(v^L|v^R)$ (on admet que $v^L<v^R$ si $v^L$ et $v^R$ sont $\neq\emptyset$)

**Théorème**
- si $v=0$, le joueur qui joue en second, que ce soit Left ou Right, a une stratégie gagnante,
- si $v>0$, Left a une stratégie gagnante qu'il joue en premier ou en second,
- si $v<0$, Right a une stratégie gagnante qu'il joue en premier ou en second

Ce théorème de la [*théorie combinatoire des jeux*](https://bobson.ludost.net/copycrime/Winning%20Ways%202nd%20Edition/Winning.Ways.for.Your.Mathematical.Plays.V1--1568811306.pdf) (voir un résumé [ici](cgt.pdf)), est admis. $\blacksquare$



Le jeu parfait pour Left (resp. Right), et en particulier la stratégie gagnante s'il en a une, est de toujours jouer le coup pour lequel la valeur de 
la position obtenue soit maximum (resp. minimum).

### Somme de deux positions

On définit la *somme* $G_1+G_2$ de deux positions comme la position obtenue en disposant côte à côte les figures de $G_1$ et de $G_2$.

**Théorème**
$V(G_1+G_2)=V(G_1)+V(G_2)$

Admis $\blacksquare$

Le calcul de la valeur d'une position nécessite, en principe, le calcul des valeurs de toutes les sous positions ; la complexité de ce calcul est exponentielle en le nombre d'arêtes. Le théorème précédent permet de réduire drastiquement ces calculs quand on peut décomposer la position en somme de plusieurs positions plus simples.

Ainsi, si $G_1$ désigne la figure précédente, de valeur $1/2$, et si $G_2$ est la *tige* 

![stack](images/stack.png)

de valeur $-3/2$ (exercice), alors $G_1+G_2$ est la figure

![dog3](images/dog3.png)

et sa valeur est $1/2+(-3/2)=-1$.


### Implémentation

Limitations : le programme suivant ne traite que des figures dont les arêtes sont graphiquement rectilignes, ce qui empêche de traiter les figures contenant des boucles ou des cycles de longueur 2.

```python
import math
import json
import drawSvg as draw
from drawSvg.widgets import DrawingWidget
from copy import deepcopy

xmax, ymax, eps, leftMargin, epsilon = 1200, 520, 30, 100, 20 # ne pas modifier ymax - eps
nGrid, topMargin = (ymax - eps) // epsilon, (ymax - eps) % epsilon

class HACKENBUSH():

    def __init__(self, name = None):

        if name is None:
            self.VERTICES = []
            self.EDGES = []
        else:
            self.load(name)

    def load(self, name):
     
        file = f'exemplesRBhackenbush/{name}.json'
        with open(file) as f:
            d = json.load(f)  
            self.VERTICES = d['vertices']
            self.n = len(self.VERTICES)
            self.EDGES = d['edges']
            self.begin = True
           
        self.init()

    def save(self, name):
        
        d = {'vertices': self.VERTICES,
             'edges': self.EDGES}
        
        file = f'exemplesRBhackenbush/{name}.json'
        with open(file, 'w') as f:
            json.dump(d, f)

    def init(self):

        self.vertex = list(map(tuple, self.VERTICES))
        self.index_of_vertex = {self.vertex[i] : i for i in range(len(self.vertex))}
        self.vertices = set(range(len(self.vertex))) # ensemble des indices des sommets non encore effacés
        self.ground = {i for i in range(len(self.vertex)) if self.vertex[i][1] == ymax - eps}
        
        def toTuple(edge):
            e, c = edge
            return tuple(e), c
        
        self.edge = list(map(toTuple, self.EDGES))
        self.index_of_edge = {self.edge[i] : i for i in range(len(self.edge))}
        self.edges = set(range(len(self.edge))) #  ensemble des indices des aretes non encore effacées

        # Calcul des composantes de la figure
        # ce qui revient a calculer les composantes connexes du graphe obtenu en supprimant
        # les aretes dont une extrémité est sur le sol
        NGvertices = {i for i in self.vertices if not(i in self.ground)} # sommets hors sol
        NGedges = {j for j in self.edges}                                # aretes hors sol
        for j in self.edges:
            (x1, y1, x2, y2), _ = self.edge[j]
            if self.index_of_vertex[(x1, y1)] in self.ground or self.index_of_vertex[(x2, y2)] in self.ground:
                NGedges -= {j}
        self.CCedges = []    # liste des composantes (ensembles d'aretes)
        mark = {k : False for k in range(self.n)}
        for i in NGvertices:
            if not mark[i]:
                Cvertex = self.cc(i, NGedges) # une composante connexe (ensemble de sommets hors sol)
                for k in list(Cvertex):
                    mark[k] = True
                    Cedge = set() # la composante correspondante (ensemble d'aretes, y compris celles qui touchent le sol)
                    for j in self.edges:
                        (x1, y1, x2, y2), _ = self.edge[j]
                        i1 = self.index_of_vertex[(x1, y1)]
                        i2 = self.index_of_vertex[(x2, y2)]
                        if i1 in Cvertex or i2 in Cvertex:
                            Cedge |= {j}
                self.CCedges.append(Cedge)
        self.dicValues = [{} for _ in range(len(self.CCedges))]

    def tupleOfSet(self, E):
        return tuple(sorted(list(E)))

    def Cvalue(self, c, e): 
        """
        c est l'indice d'une composante, 
        e = {i1,..,ik} est inclu dans la c-ieme composante self.CCedges[c] ;
        renvoie la valeur de e dans la c-ieme composante, independament des autres composantes
        """
        
        E = self.tupleOfSet(e)
        if E in self.dicValues[c]:
            return self.dicValues[c][E]
        else:
            vBlue = None
            vRed = None
            for i in e:                
                _, color = self.edge[i]
                ei = self.playEdgeGeneral(e, i)
                vi = self.Cvalue(c, ei)
                if color == 'blue':
                    if vBlue is None or vi > vBlue:
                        vBlue = vi
                else:
                    if vRed is None or vi < vRed:
                        vRed = vi
            v = V(vBlue,vRed)
            self.dicValues[c][E] = v
            return v

    def value(self, e):
        """
        renvoie la valeur de e, partie de self.edges.
        """
        return sum(self.Cvalue(c, e & ec) for c, ec in enumerate(self.CCedges))

    def cc(self, i, E):
        """
        renvoie la composante connexe (ensemble de sommets) du sommet i
        dans le graphe constitué des aretes de l'ensemble E
        """

        mark = {k : False for k in range(self.n)}
        def explore(k):
            if not mark[k]:
                mark[k] = True
                x, y = self.vertex[k]
                for l in E:
                    (x1, y1, x2, y2), _ = self.edge[l]
                    if (x1, y1) == (x, y):
                        explore(self.index_of_vertex[(x2, y2)])
                    elif (x2, y2) == (x, y):
                        explore(self.index_of_vertex[(x1, y1)])
                
        explore(i)
        return {i for i in range(self.n) if mark[i]}

    def playEdgeGeneral(self, e, j):  
        """
        e est un ensemble d'aretes reliees au sol, e contient j ;
        renvoie l'ensemble e privé de j et de la charge de j
        """

        E = e - {j}  
        
        (x1, y1, x2, y2), _ = self.edge[j]
        if (x1, y1) != (x2, y2):
            i1 = self.index_of_vertex[(x1, y1)]
            i2 = self.index_of_vertex[(x2, y2)]
            s1 = self.cc(i1, E)
            s2 = self.cc(i2, E)
            for _ in range(2):
                i1, i2, s1, s2 = i2, i1, s2, s1
                if s1 & self.ground and not (s2 & self.ground):
                    for i in s2:
                        x, y = self.vertex[i]
                        for k in list(E):
                            (x1, y1, x2, y2), _ = self.edge[k]
                            if (x1, y1) == (x, y) or (x2, y2) == (x, y):
                                E -= {k}
        return E
                        


    def modify(self):
        """
        Construction d'une figure
        """

        d = draw.Drawing(xmax + leftMargin, ymax)
        d.append(draw.Line(0, ymax - eps, xmax, ymax - eps, stroke_width = 1, stroke='black', stroke_dasharray='5,2'))
        
        group = draw.Group()
        d.append(group)

        last_edge = False
        a = None
        b = None

        lastChanges = []
        mode = 'blue'
        grid = True

        def drawButton(rectangle, text, extText = '', color = 'blue'):
            x, y, w, h = rectangle
            group.draw(draw.Text(text, 12, x + w // 2, y + h // 2, stroke = color, center = True, stroke_width=.3))
            group.draw(draw.Text(extText, 12, x + 2 * w, y + h // 2, stroke = color, center = True, stroke_width=.3))
            group.draw(draw.Rectangle(x, y, w, h, fill = 'none', stroke = color, stroke_width = .3))
            
        width = (leftMargin - eps) // 2
        undoButton = (xmax, eps, width, eps)
        modeButton = (xmax, 3 * eps, width, eps)
        gridButton = (xmax, 5 * eps, width, eps)

        def edgesAppend(edge):
            (x1, y1, x2, y2), c = edge
            e = None
            if x1 < x2 or (x1 == x2 and y1 < y2):
                e = (x1, y1, x2, y2)            
            elif x1 != x2 or y1 != y2:
                e = (x2, y2, x1, y1)
            if e is not None:
                self.EDGES.append((e, c))
            return (e, c)
        
        def redraw(e):
            nonlocal lastChanges, mode, grid

            group.children.clear()

            if lastChanges:
                drawButton(undoButton, 'Undo')
            drawButton(modeButton, 'Mode', extText = 'BLUE' if mode == 'blue' else 'RED')
            drawButton(gridButton, 'Grille', extText = 'Oui' if grid else 'Non')

            if grid:
                for i in range((ymax - eps - topMargin) // epsilon):
                    group.draw(draw.Line(0, topMargin + epsilon * i, xmax - eps, topMargin + epsilon * i, stroke_width = .1, stroke='black'))
                for j in range((xmax - eps) // epsilon + 1):
                    group.draw(draw.Line(epsilon * j, topMargin, epsilon * j, ymax - eps, stroke_width = .1, stroke='black'))

            for (x1, y1, x2, y2), c in e:
                line = draw.Line(x1, y1, x2, y2, stroke = c, stroke_width = 3)
                group.draw(line)
            for x, y in self.VERTICES:
                group.draw(draw.Circle(x, y, 4, stroke = 'black', fill='white'))                
                
        redraw(self.EDGES)

        def drawCoords(x, y):
            X = round(x / epsilon) if grid else x
            Y = round((y)/ epsilon) if grid else y
            group.draw(draw.Rectangle(xmax, 6.5 * eps, leftMargin, ymax, fill = 'white', stroke = 'none'))
            group.draw(draw.Text(f'{X:}', 12, xmax + eps, 7 * eps, stroke = 'black', stroke_width=.3, text_anchor='end'))
            group.draw(draw.Text(f'{Y}', 12, xmax + eps, 7.5 * eps, stroke = 'black', stroke_width=.3, text_anchor='end'))

        widget = DrawingWidget(d)

        def forceGrid(u, coord):
            if coord == 0: # u = x
                return epsilon * round(u / epsilon) if grid else u
            else: # u = y
                return topMargin + epsilon * round((u - topMargin)/ epsilon) if grid else u

        @widget.mousedown
        def mousedown(widget, x, y, info):
            nonlocal a, b, last_edge, lastChanges, mode, grid

            if x >= xmax:
                if x <= xmax + width:
                    if eps <= -y <= 2 * eps: # cancel last change
                        for change in lastChanges:
                            name = change[0]
                            data = change[1]
                            if name == 'verticesMinus':
                                del self.VERTICES[self.VERTICES.index(data)]
                            elif name == 'edgesPlus':
                                self.EDGES.append(data)
                            elif name == 'edgesMinus':
                                del self.EDGES[self.EDGES.index(data)]
                        lastChanges =[]
                    elif 3 * eps <= -y <= 4 * eps:
                        if mode == 'blue':
                            mode = 'red'
                        else:
                            mode = 'blue'
                    elif 5 * eps <= -y <= 6 * eps:
                        grid = not grid
            else:
                lastChanges =[]
                onpoint = False
                for u, v in self.VERTICES:
                    if (x - u) ** 2 + (y + v) ** 2 < 100:
                        newa, newb = u, v
                        onpoint = True
                        break
                if not onpoint:
                    if -y > ymax - eps - 7:
                        newa, newb = forceGrid(x, 0), ymax - eps
                    else:
                        d0 = 200
                        for i, ((x1, y1, x2, y2), c) in enumerate(self.EDGES):
                            if (x - x1) * (x - x2) + (y + y1) * (y + y2) < 0:
                                u, v, w = y1 - y2, x2 - x1, x1 * y2 - x2 * y1
                                d = (u * x - v * y + w) ** 2 / (u ** 2 + v ** 2)
                                if d < d0:
                                    i0, x10, y10, x20, y20, u0, v0, w0, d0, c0 = i, x1, y1, x2, y2, u, v, w, d, c
                        if d0 < 100:   
                            if abs(u0) < abs(v0):
                                newa, newb = x, round(- (u0 * x + w0) / v0)
                            else:
                                newa, newb = round((v0 * y - w0) / u0), -y
                            del self.EDGES[i0]
                            lastChanges.append(('edgesPlus', ((x10, y10, x20, y20), c0)))
                            e = edgesAppend(((x10, y10, newa, newb), c0))
                            if e is not None:
                                lastChanges.append(('edgesMinus', e))
                            e = edgesAppend(((x20, y20, newa, newb), c0)) 
                            if e is not None:
                                lastChanges.append(('edgesMinus', e))
                        else:
                            newa, newb = forceGrid(x, 0), forceGrid(-y, 1)
                if last_edge:
                    e = edgesAppend(((a, b, newa, newb), mode))
                    if e is not None:
                        lastChanges.append(('edgesMinus', e))
                    last_edge = False
                if not onpoint:
                    self.VERTICES.append((newa, newb))
                    lastChanges.append(('verticesMinus', (newa, newb)))
                a, b = newa, newb
                last_edge = False

            redraw(self.EDGES)
            drawCoords(x, -y)
            widget.refresh()
            try:
                self.init()
            except:
                pass

        @widget.mousemove
        def mousemove(widget, x, y, info):
            nonlocal a, b, last_edge
            if a is not None and (info['shiftKey'] or info['ctrlKey']):
                last_edge = True
                redraw(self.EDGES + [((a, b, x, -y), mode)])
            drawCoords(x, -y)
            widget.refresh()

        return widget

    
    def playEdge(self, j): 

        self.edges = self.playEdgeGeneral(self.edges, j)
        self.vertices = set()
        for i in self.edges:
            (x1, y1, x2, y2), _ = self.edge[i]
            self.vertices |= {self.index_of_vertex[(x1, y1)], self.index_of_vertex[(x2, y2)]}                              

    def play(self, hints = 2):
        """
        hints = 0 : aucune indication
        hints = 1 : indication de la valeur de la figure
        hints = 2 : de plus, indications des valeurs des sous figures 
        """
        
        self.init()

        maxX = 1
        for x, y in self.VERTICES:
            maxX = max(maxX, x)
        maxX += 50

        d = draw.Drawing(maxX, ymax)
        d.append(draw.Line(0, ymax - eps, maxX, ymax - eps, stroke_width = 1, stroke='black', stroke_dasharray='5,2'))
        Vertices_ = None
        Edges_ = None
        
        group = draw.Group()
        d.append(group)

        def play_(j):

            nonlocal Vertices_, Edges_
            
            Vertices_ = deepcopy(self.vertices)
            Edges_ = deepcopy(self.edges)
            self.playEdge(j)
            redraw()
            widget.refresh()

        def cancel():

            nonlocal Vertices_, Edges_
      
            if Vertices_ is not None:
                self.vertices = deepcopy(Vertices_)
                Vertices_ = None
                self.edges = deepcopy(Edges_)  
                redraw()
                widget.refresh()    

        def redraw():

            group.children.clear()

            for j in self.edges:
                width = 3
                (x, y, x1, y1), c = self.edge[j]
                group.draw(draw.Line(x, y, x1, y1, stroke=c, stroke_width=width))
            for i in self.vertices:
                x, y = self.vertex[i]
                group.draw(draw.Circle(x, y, 4, stroke = 'black', fill='white'))
            if hints >= 1:
                v = self.value(self.edges)
                l = [self.vertex[i][0] for i in self.ground]
                x = sum(l) // len(l)
                if self.begin:
                    s = f'{v}   ({sum(len(d) for d in self.dicValues)} positions calculées)'
                    self.begin = False
                else:
                    s = f'{v}'                
                group.draw(draw.Text(s, 12, x, 510, stroke = 'black'))
                if hints == 2:
                    for i in self.edges:
                        (x1, y1, x2, y2), c = self.edge[i]
                        vi = self.value(self.playEdgeGeneral(self.edges, i))
                        group.draw(draw.Text(f'{vi}', 14, (x1 + x2) // 2, (y1 + y2) // 2, stroke = 'red' if c == 'blue' else 'blue'))

        redraw()

        widget = DrawingWidget(d)

        @widget.mousedown
        def mousedown(widget, x, y, info):
            esc = True
            d0 = 200
            for j in self.edges:
                (x1, y1, x2, y2), _ = self.edge[j]
                if (x - x1) * (x - x2) + (y + y1) * (y + y2) < 0:
                    u, v, w = y1 - y2, x2 - x1, x1 * y2 - x2 * y1
                    d = (u * x - v * y + w) ** 2 / (u ** 2 + v ** 2)
                    if d < d0:
                        j0, d0 = j, d
            if d0 < 100: 
                play_(j0)
            else:
                cancel()
                
        return widget
```

## Compléments

Le programme précédent exploite le théorème $V(G_1+G_2)=V(G_1)+V(G_2)$ mais il existe d'autres techniques permettant de simplifier le calcul de la valeur d'une position.  
On se propose d'en établir quelques unes (avec les démonstrations cette fois).

### Opposé d'un jeu

**Propriété**
Si $G$ est une position quelconque de valeur $v$, la position $-G$ obtenue en changeant la couleur de chaque arête de $G$ a pour valeur $-v$.

**preuve**
On  montre que $V(G+(-G))=V(G)+V(-G)=0$ ou encore que, dans le jeu $G+(-G)$, le joueur 2 (qui joue en deuxième) a une stratégie gagnante :  
Si le joueur 1 joue une arête dans $G$ ou dans $-G$; le joueur 2 n'a qu'à jouer l'arête correspondante dans $-G$ ou $G$. $\blacksquare$


### Valeur d'une tige

On se propose de calculer la valeur d'une tige $s$ comme, par exemple 

![stack](images/stack.png)

notée $s=RRB$.

Considérons une tige $s$ de longueur (nombre d'arêtes) non nulle dont on cherche à calculer la valeur.  
Comme $V(s)=-V(-s)$, on peut supposer que la première arête de $P$ est bleue.


Left a une stratégie gagnante (qu'il joue en premier ou en deuxième), qui consiste à jouer la première arête. La valeur $V(s)$ est donc $>0$.

Soit $n$ le nombre de $B$ au début de l'écriture de $s$.  
Ainsi, ou bien $s=B\ldots B$ $n$ fois, ce que l'on note $s=B^n$, ou bien $s=B^nA_0\ldots A_m$ où $A_i\in\{B,R\}$ avec $A_0=R$
(on peut considérer que le premier cas correspond à $m=-1$).

**Propriété**

On pose $\sigma_i=\begin{cases}1&\text{si } A_i=B\\-1&\text{si } A_i=R\end{cases}$. Alors  
$
\displaystyle V(B^nRA_1\ldots A_m)=n+\frac{\sigma_0}{2}+\cdots+\frac{\sigma_m}{2^{m+1}}
$

**preuve**

On voit d'abord, par récurrence, que $V(B^n)=n$ : $V(B^0)=(|)=0$ et si $V(B^i)=i$ pour $i<n$, alors $V(B^n)=(\max\{0,\ldots,n-1\}|)=(n-1|)=n$.

On note $\mathcal V^L(s)$ (resp. $\mathcal V^R(s)$) l'ensemble des valeurs des tiges obtenues en supprimant une arête bleue (resp. rouge) dans $s$ ;  
et $v^L(s):=\max\mathcal V^L(s)$, $v^R(s):=\min\mathcal V^R(s)$, de sorte que $V(s)=(v^L(s)|v^R(s))$.

$n$ étant fixé, on procède alors par récurrence sur $m$ :

On suppose le résultat acquis pour $s_i=B^nRA_1\ldots A_i$, $i=-1,0,\ldots, m$ et on le montre pour $s_mB$ (si $m\geqslant0$) et pour $s_mR$ (si $m\geqslant -1$).  
On introduit $p$ tq $\displaystyle V(s_m)=n+\frac{\sigma_0}{2}+\cdots+\frac{\sigma_m}{2^{m+1}}=\frac{p}{2^{m+1}}$
($p\in\mathbf N$ est impair)
- Calcul de  $V(s_mB)=(v^L(s_mB)|v^R(s_mB))$ :
  - On a $\mathcal V^L(s_mB)=\mathcal V^L(s_m)\cup\{V(s_m)\}$, or  $V(s_m)=(v^L(s_m)|v^R(s_m))>v^L(s_m)=\max\mathcal V^L(s_m)$, donc   
    $\displaystyle v^L(s_mB)=\max(\mathcal V^L(s_m)\cup \{V(s_m)\})=V(s_m)=\frac{p}{2^{m+1}}$.  
  - D'autre part, $\mathcal V^R(s_mB)=\mathcal V^R(s_m)$ donc $\displaystyle v^R(s_mB)=v^R(s_m)=\frac{p+1}{2^{m+1}}$. En effet :
    - si $\displaystyle v^R(s_m)>\frac{p+1}{2^{m+1}}$ alors $\displaystyle \frac{p+1}{2^{m+1}}\in]v^L(s_m),v^R(s_m)[$,  
      mais, comme $p+1$ est pair, $\displaystyle \frac{p+1}{2^{m+1}}$ a un dénominateur strictement inférieur
      à celui de $\displaystyle \frac{p}{2^{m+1}}=V(s_m)=(v^L(s_m)|v^R(s_m))$ ce qui est impossible ;
    - et si $\displaystyle v^R(s_m)<\frac{p+1}{2^{m+1}}$ alors  $\displaystyle\frac{p}{2^{m+1}}< v^R(s_m)<\frac{p+1}{2^{m+1}}$,  
      mais $v^R(s_m)$ a un dénominateur $\leqslant 2^m$ (hypothèse de récurrence) ce qui est clairement impossible.  

  Des deux calculs précédents, on tire   
  $\displaystyle
   V(s_mB)=(\frac{p}{2^{m+1}}|\frac{p+1}{2^{m+1}})=\frac{2p+1}{2^{m+2}} \text{ (vérification facile)}
   = \frac{p}{2^{m+1}}+\frac{1}{2^{m+2}}=V(s_m)+\frac{1}{2^{m+2}}=n+\frac{\sigma_0}{2}+\cdots+\frac{\sigma_m}{2^{m+1}}+\frac{1}{2^{m+2}}.
  $
- On montre de même que $\displaystyle V(s_mR)=n+\frac{\sigma_0}{2}+\cdots+\frac{\sigma_m}{2^{m+1}}-\frac{1}{2^{m+2}}$. $\blacksquare$ 

**Autre formule**

Si $n>0$ et $m\geqslant0$, on pose $a_i=\begin{cases}1&\text{si } A_i=B\\0&\text{si } A_i=R\end{cases}$. Alors  
$V(B^nRA_1\ldots A_m)=n-1+\overline{0,a_1\ldots a_m1}$ 

où $\overline{0,a_1\ldots a_m1}:=a_12^{-1}+\cdots+ a_m2^{-m}+ 2^{-m-1}$ (écriture en base $2$).

**preuve** exercice $\blacksquare$

**Corollaire**

La fonction $V$ induit une bijection de l'ensemble des tiges sur $\mathbf D$.

### Implémentation

On représente une tige par une chaine des caractères 🟦 et 🟥, par exemple $\texttt{'🟥🟥🟦'}$

```python
Blue = '🟦'
Red = '🟥'

def valueOfStalk(s):
    if s:
        if s[0] == Blue:
            if Red in s:
                n = s.index(Red)
                return n - 1 + int(s[n + 1:].replace(Blue,'1').replace(Red, '0') + '1', 2) / S(2) ** (len(s) - n)
            else:
                return S(len(s))
        else:
            return - valueOfStalk(s.replace(Red, '.').replace(Blue,Red).replace('.',Blue))
    else:
        return S(0)
    
def stalkOfValue(x):
    if x:
        if x > 0:
            if denom(x) == 1:
                return Blue * x
            else:
                n = floor(x) + 1
                y = x - n + 1
                m = log(denom(y), 2) - 1
                t = bin(numer(y))[2:-1].replace('1',Blue).replace('0',Red)
                return Blue * n + Red * (m - len(t) + 1) + t
        else:
            return stalkOfValue(- x).replace(Red, '.').replace(Blue,Red).replace('.',Blue)
    else:
        return ''
```

### Le principe du colon

Soient $G$ et $H$ deux figures et $s$ un sommet de $G$.  
Dans $H$, on peut identifier tous les sommets du sol sans rien changer ; on  suppose donc que le sol de $H$ est réduit à un seul sommet $s_H$.   
Si l'on accroche $H$ à $G$ en identifiant $s$ et $s_H$, on obtient une nouvelle figure notée $G_s:H$.

![Gs](images/Gs.png) $\qquad$  ![H](images/H.png) $\qquad$  ![GsH](images/GsH.png)

**Principe du colon**

Si $G$, $H$, $K$ sont trois figures et $s$ est un sommet de $G$, alors

$V(H)\leqslant V(K)\Rightarrow V(G_s:H)\leqslant V(G_s:K)$ et donc

$V(H)= V(K)\Rightarrow V(G_s:H)= V(G_s:K)$

**preuve**

On note $K-H:=K+(-H)$.

$V(H)\leqslant V(K)$ équivaut à $V(K-H)=V(K)-V(H)\geqslant0$ qui équivaut à dire que, si, dans le jeu $K-H$,
Left est le joueur 2, il a une stratégie gagnante. Faisons donc l'hypothèse que c'est le cas et montrons qu'alors,
c'est aussi le cas pour le jeu $G_s:K-G_s:H$ :

Considérons donc le jeu $G_s:K-G_s:H$. 
- Si Right (joueur 1) joue une arête dans $K$ ou dans $-H$, on sait, vue l'hypothèse,
que, dans $K-H$, Left a une réponse qui donne un jeu $K'-H'$ de valeur $\geqslant0$. Dans
le jeu $G_s:K-G_s:H$, Left joue $G_s:K'-G_s:H'$.
- Si Right joue une arête dans la composante $G$ de $G_s:K$ ou $-G_s:H$, mettons $G_s:K$, alors Left répond en jouant
l'arête correspondante dans la composante $G$ de $-G_s:H$. Il se peut que la charge de l'arête jouée par Right
contienne $K$ ; alors la charge de l'arête jouée par Left contiendra $H$ et le jeu obtenu aura la forme $G'-G'$, où $G'$
est une sous figure de $G$,  jusqu'à la fin
de la partie. $\blacksquare$
